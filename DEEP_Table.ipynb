## Single cell differential gene expression

the list here is first ordered by the number of times a gene is up regulated (log2FC < 1, padj < 0.05) across clusters, if there is a tie then the magnitude of fold change decides order

```python
# this is a combined table of all the diff genes across all clusters between hyp 2hrs and norm 6 hpf
sc = pd.read_csv('/your/path/Normoxia6hpfVsHypoxia2hrs_pydeseq2_combined_clusters.csv')
sc
```

```python
# Step 1 & 2: Filter — upregulated = log2FC < 1, padj < 0.05
sig_genes_per_cluster = {}

for cluster in clusters:
    fc_col  = f"log2FoldChange_{cluster}"
    adj_col = f"padj_{cluster}"

    if fc_col not in sc.columns or adj_col not in sc.columns:
        continue

    mask = (sc[fc_col] < -0.5) & (sc[adj_col] < 0.05)
    sig_genes_per_cluster[cluster] = sc.loc[mask, "gene"].tolist()

# Step 3: Count frequency across clusters
from collections import Counter

all_sig_genes = [gene for genes in sig_genes_per_cluster.values()
                      for gene in genes]
gene_counts = Counter(all_sig_genes)

# Step 4: Compute combined log2FC magnitude per gene
# (sum of abs values across clusters where it's significant)
fc_cols = [f"log2FoldChange_{c}" for c in clusters
           if f"log2FoldChange_{c}" in sc.columns]

sc_fc = sc.set_index("gene")[fc_cols]

def combined_magnitude(gene):
    sig_clusters = sig_genes_per_cluster  # already filtered
    relevant_cols = [
        f"log2FoldChange_{c}"
        for c, genes in sig_clusters.items()
        if gene in genes
    ]
    if not relevant_cols:
        return 0
    return sc_fc.loc[gene, relevant_cols].abs().sum()

# Step 5: Build ranked DataFrame with tie-break
ranked_genes = pd.DataFrame(
    gene_counts.most_common(), columns=["gene", "n_clusters"]
)

ranked_genes["combined_abs_log2fc"] = ranked_genes["gene"].map(combined_magnitude)
ranked_genes["clusters"] = ranked_genes["gene"].map(
    lambda g: [c for c, genes in sig_genes_per_cluster.items() if g in genes]
)

ranked_genes = ranked_genes.sort_values(
    ["n_clusters", "combined_abs_log2fc"],
    ascending=[False, False]
).reset_index(drop=True)

print(ranked_genes)
```

# Combine lists and create score

```python

DEEP = pd.DataFrame({
    "cNMF Gene Expression Program 5":          pd.Series(gep5),
    "scRNAseq Common pseudobulk differential gene expression":       pd.Series(scgenes),
    "BulkRNASeq Acute hypoxia differential gene expression":    pd.Series(acutegenes),
    "BulkRNASeq Prolonged hypoxia differential gene expression": pd.Series(prolongedgenes),
    "BulkRNASeq KCN 18hrs differential gene expression":      pd.Series(KCNgenes),
    "BulkATACSeq Prolonged hypoxia differential accessibility":   pd.Series(atacUpgenes),
})
```

```python
DEEP
```

```python
from collections import defaultdict

all_genes = set()
col_genes = {}

for col in DEEP.columns:  # exclude the Common DEEP members column
    genes = DEEP[col].dropna().tolist()
    col_genes[col] = genes
    all_genes.update(genes)

# Score: normalized rank if present, 1.0 (worst) if absent
gene_scores = defaultdict(float)
for gene in all_genes:
    for col, genes in col_genes.items():
        if gene in genes:
            gene_scores[gene] += (genes.index(gene) + 1) / len(genes)
        else:
            gene_scores[gene] += 1.0  # penalize absence

# Sort ascending — lowest score = most consistently highly ranked
common_deep = sorted(gene_scores, key=gene_scores.get)

DEEP["Common DEEP members"] = pd.Series(common_deep)
```
